```python
DATA_YAML = "/kaggle/input/fishclassification-dataset-ma/dataset.yaml"
MODEL = "yolov8n.pt"
```

```python
EPOCHS = 20       # Lower for faster experimentation, increase for better accuracy
IMG_SIZE = 416    # Smaller size = faster, but less accurate (~416 or 512 for speed)
BATCH = 32   
```

```python
model = YOLO(MODEL)
model.train(
    data=DATA_YAML,
    epochs=EPOCHS,
    imgsz=IMG_SIZE,
    batch=BATCH,
    device=0,             # 0 for first GPU
    workers=2,            # Fewer workers = less CPU/RAM usage, tweak as needed
    project="fish_yolov8_fast",
    name="exp",
    exist_ok=True,
    verbose=True
)
```
